<h2>Time Series Visualization</h2>

```python
plt.figure(figsize=(12, 5))
plt.plot(pems_bay.iloc[:, 0], label="PEMS-BAY Sensor 1", alpha=0.7)
plt.plot(metr_la.iloc[:, 0], label="METR-LA Sensor 1", alpha=0.7)
plt.legend()
plt.title("Traffic Time Series - Sample Sensors")
plt.xlabel("Time")
plt.ylabel("Traffic Volume")
plt.show()
```

**Traffic Volume Patterns**:

- The METR-LA dataset shows a relatively consistent traffic volume in its recorded period, with values clustering around the same range.
- The PEMS-BAY dataset exhibits more variability, with visible fluctuations in traffic volume over time.
The difference in variance suggests possible differences in traffic congestion patterns between the two locations.

<h2>Autocorrelation Plot for PEMS BAY</h2>

```python
from pandas.plotting import autocorrelation_plot

autocorrelation_plot(pems_bay.iloc[:, 0])  
plt.show()
```

- **High autocorrelation at small lags**: Values start around 0.75 at lag 0 and gradually decrease, indicating strong short-term correlation where recent traffic significantly influences upcoming values.
- **Periodic spikes**: Repeating peaks at regular intervals suggest seasonality, likely reflecting daily patterns like rush hours and nighttime lows.
- **Slow decay**: Autocorrelation declines gradually, showing long-term dependency, meaning past traffic conditions influence future conditions over extended periods.
- **Noise at higher lags**: Beyond ~30,000 lags, autocorrelation fluctuates at low values, indicating weakened long-term relationships but persistent periodic effects.

<h2>Autocorrelation Plot for METR LA</h2>

```python
from pandas.plotting import autocorrelation_plot

autocorrelation_plot(metr_la.iloc[:, 0])  
plt.show()
```

- **Weak short-term correlation**: Autocorrelation starts around 0.3–0.4 at lag 0, lower than PEMS-BAY, indicating weaker short-term dependencies in LA traffic, likely due to higher randomness from congestion, accidents, or weather.  
- **Rapid decay**: Autocorrelation drops to nearly zero within a few thousand lags, suggesting past traffic conditions have minimal influence on future traffic beyond a short period, emphasizing the importance of short-term forecasting.  
- **No strong periodicity**: Unlike PEMS-BAY, METR-LA lacks clear seasonal peaks, indicating more unpredictable and less cyclical traffic, possibly due to irregular urban congestion rather than structured daily patterns.  
- **Medium-range fluctuations**: Small oscillations beyond lag ~10,000–20,000 suggest weak long-term dependencies, potentially reflecting subtle weekly or weekend patterns, though less pronounced than in PEMS-BAY.

<h2>Traffic Volume Distribution</h2>

```python
plt.figure(figsize=(12, 5))
sns.kdeplot(pems_bay.iloc[:, 0], label="PEMS-BAY", fill=True, alpha=0.6)
sns.kdeplot(metr_la.iloc[:, 0], label="METR-LA", fill=True, alpha=0.6)
plt.legend()
plt.title("Traffic Volume Distribution")
plt.xlabel("Traffic Volume")
plt.show()
```

- **Peak Traffic Differences**:
  - PEMS-BAY peaks sharply around 75-80, indicating concentrated high traffic.  
  - METR-LA peaks broadly around 60-70, showing more variability.  


- **Traffic Spread**:  
  - METR-LA ranges from near 0 to 70+, with a mix of low and high volumes.  
  - PEMS-BAY is narrower, mostly in a high range.  


- **Low Traffic in METR-LA**:  
  - METR-LA has significant low volumes (0-20), absent in PEMS-BAY, suggesting off-peak periods or less congested roads.  


- **Overlap**:  
  - Both datasets overlap at 60-70, but PEMS-BAY dominates beyond 70, indicating heavier congestion.

<h2>Principal Component Analysis and Clustering</h2>

<h3>For PEMS BAY</h3>

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Normalize data before PCA
scaler = StandardScaler()
pems_bay_scaled = scaler.fit_transform(pems_bay)
```

```python
pca = PCA(n_components=2)  # Reduce to 2D for visualization
pems_bay_pca = pca.fit_transform(pems_bay_scaled)

# Scatter plot of PCA results
plt.scatter(pems_bay_pca[:, 0], pems_bay_pca[:, 1], alpha=0.5)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Projection of Sensor Data")
plt.show()
```

```python
from sklearn.cluster import KMeans

# Apply KMeans clustering
kmeans = KMeans(n_clusters=5, random_state=42)  # Choose appropriate k
pems_bay["cluster"] = kmeans.fit_predict(pems_bay_scaled)

# Scatter plot using PCA components
plt.scatter(pems_bay_pca[:, 0], pems_bay_pca[:, 1], c=pems_bay["cluster"], cmap="viridis", alpha=0.6)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Clustering Sensors Using K-Means")
plt.colorbar(label="Cluster")
plt.show()
```

<h3>For METR LA</h3>

```python
scaler = StandardScaler()
metr_la_scaled = scaler.fit_transform(metr_la)
```

```python
metr_la_pca = pca.fit_transform(metr_la_scaled)

# Scatter plot of PCA results
plt.scatter(metr_la_pca[:, 0], metr_la_pca[:, 1], alpha=0.5)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Projection of Sensor Data")
plt.show()
```

```python
metr_la["cluster"] = kmeans.fit_predict(pems_bay_scaled)

# Scatter plot using PCA components
plt.scatter(metr_la_pca[:, 0], metr_la_pca[:, 1], c=metr_la["cluster"], cmap="viridis", alpha=0.6)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Clustering Sensors Using K-Means")
plt.colorbar(label="Cluster")
plt.show()
```
